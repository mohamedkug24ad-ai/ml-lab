```python
# Step 1: Import required libraries
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# Step 2: Create a text dataset with labels

texts = [
    "The team won the football match",
    "The player scored a goal",
    "Cricket is a popular sport",
    "The football tournament starts today",
    "The batsman scored a century",
    "The team practiced for the cricket match",

    "The new smartphone has a powerful processor",
    "Artificial intelligence is changing technology",
    "Python is used for machine learning",
    "The computer has a fast graphics card",
    "Machine learning is a branch of artificial intelligence",
    "The latest laptop has advanced technology"
]

labels = [
    "Sports",
    "Sports",
    "Sports",
    "Sports",
    "Sports",
    "Sports",

    "Technology",
    "Technology",
    "Technology",
    "Technology",
    "Technology",
    "Technology"
]

# Step 3: Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    texts,
    labels,
    test_size=0.25,
    random_state=42,
    stratify=labels
)

# Step 4: Convert text data into TF-IDF vectors

vectorizer = TfidfVectorizer()

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# Step 5: Train the SVM classifier

model = SVC(kernel='linear')
model.fit(X_train_tfidf, y_train)

# Step 6: Predict categories for testing data

y_pred = model.predict(X_test_tfidf)

# Step 7: Find accuracy and display results

accuracy = accuracy_score(y_test, y_pred)

print("SVM Text Categorization")
print("-----------------------")

print("\nAccuracy:", accuracy)

print("\nActual vs Predicted:")
for actual, predicted, text in zip(y_test, y_pred, X_test):
    print("\nText:", text)
    print("Actual Category:", actual)
    print("Predicted Category:", predicted)

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
